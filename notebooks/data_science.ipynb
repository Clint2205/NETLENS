```python
import numpy as np

x = df['duration_seconds']
y = df['packets_per_second']

slope, intercept = np.polyfit(x, y, 1)

print(f"Slope: {slope}, Intercept: {intercept}")
```

```python
##predict the number of packets per second for a given duration using the linear regression equation:

def predict_packets(duration):
    return slope * duration + intercept




print(predict_packets(10))  # Predict packets per second for a duration of 10 seconds
```
